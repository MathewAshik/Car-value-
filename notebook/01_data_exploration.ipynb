```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.hist(df_clean['price'], bins=30)
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of C    ')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df_clean['price'], bins=30)

plt.title('Distribution of Used Car Prices')
plt.xlabel('Price')
plt.ylabel('Number of Cars')

plt.show()
```

```python
df_clean[['distance_travelled(kms)', 'price']].corr()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df_clean['distance_travelled(kms)'],
    df_clean['price'],
    alpha=0.5
)

plt.title('Distance Travelled vs Car Price')
plt.xlabel('Distance Travelled (km)')
plt.ylabel('Price')

plt.show()
```
